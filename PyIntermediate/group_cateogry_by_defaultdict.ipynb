### Group Category in a Dictionary

The most clean and efficient way to group items by a category into a dictionary is using collections.defaultdict. This approach avoids manual key checking, runs in optimal O(N) time complexity

```python
from collections import defaultdict

expenses = [
    {"category": "food", "amount": 250.00, "description": "Jollibee"},
    {"category": "transport", "amount": 120.00, "description": "Grab"},
    {"category": "food", "amount": 180.00, "description": "Chilli's"},
]

grouped_data = defaultdict(list)

for item in expenses:
    data = {}
    category = item["category"]
    data['amount'] = item.get('amount')
    data['description'] = item.get('description')
    grouped_data[category].append(item)

print(dict(grouped_data))
```
