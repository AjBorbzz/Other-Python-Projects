# Reusable Generators in Python

### Why Generators Matter

Python generators are useful when processing large amounts of data because they produce values one at a time instead of storing the entire result in memory.

This makes them useful for tasks such as:
- Processing large files
- Streaming data
- Parsing logs
- Performing validation
- Processing database results
- Building data pipelines

Consider a normal function:

```python
def get_numbers(): 
    return [1, 2, 3, 4, 5]
```

The entire list exists in memory.

A generator produces values lazily:

```python
def get_numbers(): 
    for number in range(1, 6): 
        yield number

```

Now the values are produced only when requested.


### Generators Are Single-Use Iterators

A generator instance can normally be consumed only once.

Example:

```python
def generate_numbers(): 
    for number in range(1, 4): 
        yield number
```

Create the generator:

```python
numbers = generate_numbers()
```

First iteration:
```python
for number in numbers: 
    print(number)
```

Output

```bash
1 
2 
3
```

## The Solution: Implement the Iterable Protocol

Python objects can participate in a for loop when they implement the iterable protocol.

The important method is:

```__iter__()```

Instead of storing a generator instance, our object creates a new generator every time ```__iter__()``` is called.

Example:
```python
class ReusableNumbers:

    def __iter__(self):
        for number in range(1, 4):
            yield number
```

Create the object:

```numbers = ReusableNumbers()```

First iteration:
```python
for number in numbers:
    print(number)
```
Output:
```bash
1
2
3
```
Second iteration:
```python
for number in numbers:
    print(number)
```
Output:
```bash
1
2
3
```
It works again.






### What Is Actually Happening?

The important distinction is:

```numbers = ReusableNumbers()```

numbers itself is not the generator.

It is an iterable object.

Whenever Python executes:

```for number in numbers:```

Python effectively calls:
```python
iter(numbers)
```
which calls:

```numbers.__iter__()```

Since ```__iter__()``` contains yield, it creates a new generator.

Conceptually:
```text
ReusableNumbers object
        |
        | iter(numbers)
        v
Generator #1
        |
     consumed
        |
      finished


ReusableNumbers object
        |
        | iter(numbers)
        v
Generator #2
        |
     consumed
        |
      finished
```
Each loop receives an independent generator.

### Example #1 : Large log or file processing

Example: SOC logs, CSV exports, application logs, transaction files.

````python
class LogReader:
    def __init__(self, filename):
        self.filename = filename

    def __iter__(self):
        with open(self.filename) as file:
            for line in file:
                yield line
````

You can make several passes without loading a 5–20 GB file into RAM:

````python
logs = LogReader("security.log")

error_count = sum("ERROR" in line for line in logs)

for line in logs:
    if "CRITICAL" in line:
        print(line)
````

Why it fits: each analysis needs the same large source, but storing the whole dataset would waste memory.

### Example # 2: Database or paginated API processing

Useful when records are retrieved in batches from a database or API.

````python
class UserRecords:
    def __init__(self, database):
        self.database = database

    def __iter__(self):
        for batch in self.database.fetch_batches(size=1000):
            yield from batch
````

Why it fits: millions of rows can be streamed rather than placed into one huge list.

Important caveat: every iteration may execute the database query or API request again.

### Example # 3: ETL and data-processing pipelines

Useful for processing large datasets through multiple transformations.

The same reusable iterable can support different operations:
